```python
doc=fitz.open(chemin_pdf)

#compter les pages du pdf
nb_pages=len(doc)
print(f"Le doc contient {nb_pages} pages")

#doc.close()
#libère les ressources système (mémoire et fichiers) que le PDF utilisait, exactement comme on ferme un fichier Word pour ne pas le laisser "ouvert" en mémoire.
```

```python
num_page=5

page=doc[num_page]

texte=page.get_text()

print(texte)
```

```python
def verifier_accent(texte):
    """
    vérifie la présence des accents dans un texte.
    retourne le nombre de caractères accentués
    """
    pattern_accent=r'[éèêëàâäôöûüçîï]'
    accents=re.findall(pattern_accent,texte.lower())
    return len(accents)

print(f"nombre de caractères accentués dans la page {num_page} est: {verifier_accent(texte)}")
```
