```python
def read_pdf(pdf_path):
    """
    Lis un pdf et retorune le texte de chaque page
    """

    pages=[]
    with fitz.open(pdf_path) as doc:
        for i in range(len(doc)):
            pages.append({
                "page": i+1,
                "text" : doc[i].get_text()
            })
        return pages
```

```python
from collections import Counter

def extraire_tous_les_textes_avec_styles(pdf_path, page_numbers=None):
    """
    Extrait TOUS les textes de chaque page avec leurs styles.
    
    Paramètres:
        pdf_path (str): Chemin vers le PDF
        page_numbers (list, optional): Liste des pages à analyser (ex: [1, 2, 3])
                                       Si None, analyse toutes les pages.
    
    Retourne:
        dict: {
            "page_1": [
                {"texte": "...", "police": "...", "taille": 11.04, "style": "normal"},
                ...
            ],
            ...
        }
    """
    
    resultats = {}
    doc = fitz.open(pdf_path)
    
    # Pages à analyser
    if page_numbers is None:
        pages_a_analyser = range(1, len(doc) + 1)
    else:
        pages_a_analyser = page_numbers
    
    for num_page in pages_a_analyser:
        # Vérifier que la page existe
        if num_page > len(doc):
            print(f"⚠️ Page {num_page} n'existe pas (max: {len(doc)})")
            continue
        
        page = doc[num_page - 1]  # 0-indexé
        data = page.get_text("dict")
        
        page_key = f"page_{num_page}"
        resultats[page_key] = []
        
        # Parcourir la structure
        for bloc in data.get("blocks", []):
            if bloc["type"] != 0:  # Ignorer les images
                continue
            
            for ligne in bloc.get("lines", []):
                # Extraire le texte complet de la ligne
                texte_ligne = ""
                for span in ligne.get("spans", []):
                    texte_ligne += span.get("text", "")
                
                if not texte_ligne.strip():
                    continue
                
                # Prendre le premier span pour les infos de style
                # (on suppose que la ligne a un style cohérent)
                span_principal = ligne.get("spans", [{}])[0]
                
                police = span_principal.get("font", "unknown")
                taille = round(span_principal.get("size", 0), 2)
                
                # Détecter le style
                style = "normal"
                if "Bold" in police and "Italic" in police:
                    style = "bolditalic"
                elif "Bold" in police:
                    style = "bold"
                elif "Italic" in police:
                    style = "italic"
                
                # Position (pour détecter l'indentation)
                bbox = span_principal.get("bbox", [0, 0, 0, 0])
                x0 = round(bbox[0], 2) if len(bbox) > 0 else 0
                
                resultats[page_key].append({
                    "texte": texte_ligne.strip(),
                    "police": police,
                    "taille": taille,
                    "style": style,
                    "x0": x0,  # Position horizontale (pour détecter l'indentation)
                    "hauteur": round(ligne.get("bbox", [0, 0, 0, 0])[3] - ligne.get("bbox", [0, 0, 0, 0])[1], 2)
                })
    
    doc.close()
    return resultats
```

```python
def afficher_extraction(resultats, pages=None):
    """
    Affiche les résultats d'extraction de manière lisible.
    
    Paramètres:
        resultats (dict): Résultat de extraire_tous_les_textes_avec_styles
        pages (list, optional): Pages à afficher (ex: [1, 2, 3])
    """
    
    if pages is None:
        pages = list(resultats.keys())
    
    for page_key in pages:
        if page_key not in resultats:
            print(f"⚠️ {page_key} non trouvé")
            continue
        
        print(f"\n{'='*70}")
        print(f"📄 {page_key}")
        print(f"{'='*70}")
        
        # Afficher un résumé
        items = resultats[page_key]
        print(f"Nombre d'éléments : {len(items)}")
        
        # Compter les styles
        styles = Counter()
        polices = Counter()
        tailles = Counter()
        
        for item in items:
            styles[item["style"]] += 1
            polices[item["police"]] += 1
            tailles[item["taille"]] += 1
        
        print(f"\n📊 Résumé des styles :")
        print(f"   Styles : {dict(styles)}")
        print(f"   Polices : {dict(polices)}")
        print(f"   Tailles : {dict(tailles)}")
        
        print(f"\n📝 Détail des lignes :")
        print("-" * 70)
        for i, item in enumerate(items[:20]):  # Limiter à 20 pour lisibilité
            texte = item["texte"]
            if len(texte) > 60:
                texte = texte[:57] + "..."
            
            print(f"{i+1:3}. {texte:<40} | {item['police']:<20} | {item['taille']:<6} | {item['style']:<10} | x={item['x0']:<8}")
        
        if len(items) > 20:
            print(f"\n   ... et {len(items) - 20} autres lignes")
        
        print("-" * 70)
```

```python
def detecter_patterns_par_niveau(resultats, seuil_stabilite=0.8):
    """
    Détecte automatiquement les patterns de style par niveau.
    
    Paramètres:
        resultats (dict): Résultat de extraire_tous_les_textes_avec_styles
        seuil_stabilite (float): Seuil pour considérer un pattern comme stable
    
    Retourne:
        dict: {
            "niveau_1": {"police": "Calibri", "taille": 14, "style": "italic", "confidence": 0.95},
            ...
        }
    """
    
    # Regrouper par (police, taille, style)
    groups = Counter()
    
    for page_key, items in resultats.items():
        for item in items:
            key = (item["police"], item["taille"], item["style"])
            groups[key] += 1
    
    # Calculer les fréquences
    total = sum(groups.values())
    
    patterns = {}
    niveau = 1
    
    # Trier par fréquence décroissante
    for (police, taille, style), count in groups.most_common():
        confidence = count / total
        
        # Ne garder que les patterns avec une confiance suffisante
        if confidence >= seuil_stabilite:
            patterns[f"niveau_{niveau}"] = {
                "police": police,
                "taille": taille,
                "style": style,
                "confidence": round(confidence, 3),
                "occurrences": count
            }
            niveau += 1
    
    return patterns
```
